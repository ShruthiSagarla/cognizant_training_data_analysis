```python
import pandas as pd

df = pd.read_excel(r"C:\Users\Dell\Documents\student.xlsx")
print(df)
```

```python
print(df.isnull())
```

```python
print(df.fillna(67,inplace=True))
```

```python
print(df['Marks'].mean())
```

```python
print(df['Marks'].sum())
```

```python
def grade(Marks):

  if Marks>=90:
   return'A'
  elif Marks>=75:
    return'B'
  elif Marks>=60:
    return'C'
  elif Marks>=50:
    return'D'

  else:
    return'fail'
df["grade"]=df["Marks"].apply(grade)

print(df)
```

```python
print(df['Marks'].median())
```

```python
print(df['Marks'].mode())
```

```python
import matplotlib.pyplot as plt
df = pd.read_excel(r"C:\Users\Dell\Documents\student.xlsx")

plt.bar(df['Student Name'],df['Marks'])
plt.title("STUDENTS REPORT")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.xticks(rotation=45)
mark_count=df['Marks'].value_counts().sort_index()
print(mark_count)
plt.show()

plt.bar(mark_count.index,mark_count.values)
plt.show()

result=mark_count.reset_index()
result.columns=["Marks","Student Count"]
print(result)
```
